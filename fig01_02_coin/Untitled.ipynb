```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.figure(dpi=300)
mu = np.linspace(0, 1, 1000)

def Bern(x, mu):
    return mu**x * (1-mu)**(1-x)

L0 = Bern(0, mu)
L1 = Bern(1, mu)
plt.plot(mu, L0, label='Tails: x=0')
plt.plot(mu, L1, label='Heads: x=1')
plt.legend()
plt.xlabel(r"$\mu$")
plt.ylabel("Likelihood")
```

```python
plt.figure(dpi=300)
plt.plot([0, 1],[1,1.], c='k', label='prior')

Z = np.trapz(L0, mu)
plt.plot(mu, L0 / Z, label='T')



Z = np.trapz(L0*L1, mu)
plt.plot(mu, L0*L1 / Z, label='TH')

Z = np.trapz(L0*L1*L1, mu)
plt.plot(mu, L0*L1*L1 / Z, label='THH')
Z = np.trapz(L0*L1**5, mu)
plt.plot(mu, L0*L1**5 / Z, label='THHHHHH')
plt.legend()
plt.grid()

plt.xlabel(r"$\mu$")
plt.ylabel("Posterior")
```

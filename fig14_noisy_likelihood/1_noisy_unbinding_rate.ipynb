# Noise in the force measurement

If the force is measured with some noise, the posterior of the unbinding-parameters is proportional to the marginalization of all possible forces, given the force measurement.

`r` is the base unbinding rate.
`x` is the unbinding length scale.
`s` is the measurement noise.
`f` is the force on the bond.
`F` is the measured force on the bond.

```python
from sympy import *
from sympy.abc import r, x, s, F, f

r, x, s, F, f, dt = symbols('r_{u,0} x s F f \delta', real=True)
```

```python
rate = dt * r * exp(abs(f) * x)
rate
```

```python
noise = 1/s * exp(-(F-f)**2 / s**2)
noise
```

```python
res = simplify(integrate(rate*noise, (f,-oo,oo)))
```

```python
import numpy as np
from numpy.random import multivariate_normal as mvn
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import axes3d
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 300
```

```python
N = 30000
SIG = np.eye(3)
MU = np.zeros(3)

pts = mvn(MU, SIG, N)
d = np.linalg.norm(pts, axis=1)

size = 5
aspect = 1.2
ax = plt.figure().add_subplot(projection='3d')
ax.scatter(*(pts*[1,1,1] + [0,0,0]).T, s=0.05, c=d, marker=',')
ax.plot(*(pts*[0,1,1] + [-size,0,0]).T, '.k', markersize=0.1, alpha=0.05, markeredgewidth=1)
ax.plot(*(pts*[1,0,1] + [0,size,0]).T, '.k', markersize=0.1, alpha=0.05)
ax.plot(*(pts*[1,1,0] + [0,0,-size]).T, '.k', markersize=0.1, alpha=0.05)
ax.set_xlim([-size,size])
ax.set_ylim([-size,size])
ax.set_zlim([-size/aspect,size/aspect])
plt.show()
```
